## Bereinigung der Spalte age:
Die Spalte enthält eine Mischung aus tatsächlichen Altersangaben und Geburtsjahren. Werte zwischen 1900 und 2002 werden daher als Geburtsjahre interpretiert und mit 2026 - Geburtsjahr in ein Alter umgerechnet. Dreistellige Werte über 100 sowie Werte über 2002 werden als unplausibel betrachtet und aus dem Datensatz entfernt.

```python
df["age"].unique()
```

```python
rows_before = len(df)

# 1. NaN-Werte und 0-Werte in age entfernen
rows_before_step1 = len(df)

df = df.dropna(subset=["age"]).copy()                 # entfernt Zeilen mit NaN in age
df = df[df["age"] != 0].copy()                       # entfernt Zeilen mit age = 0

rows_after_step1 = len(df)

# 2. Alle Werte über 2002 entfernen
rows_before_step2 = len(df)

df = df[df["age"] <= 2002].copy()                     # entfernt unrealistische Werte wie 412434 oder 999999999

rows_after_step2 = len(df)

# 3. Dreistellige Alterswerte über 100 entfernen
rows_before_step3 = len(df)

df = df[~df["age"].between(100, 999)].copy()          # entfernt z. B. 118, 188, 191, 208, 223, 266

rows_after_step3 = len(df)

# 4. Geburtsjahre von 1900 bis 2002 in Alter umwandeln
rows_before_step4 = len(df)

birthyear_mask = df["age"].between(1900, 2002)       # erkennt Werte, die Geburtsjahre sind

df.loc[birthyear_mask, "age"] = (
    2017 - df.loc[birthyear_mask, "age"]             # Geburtsjahr → Alter
)
rows_after_step4 = len(df)

df["age"].unique()                    # Kontrolle

rows_after = len(df)

print("Gelöschte Zeilen:", rows_before - rows_after)
print("Verbleibende Zeilen:", rows_after)

df["age"].unique()   # Kontrolle


# Ausgabe
print("Schritt 1 - NaN und 0 entfernt:",
      rows_before - rows_after_step1)

print("Schritt 2 - Werte > 2002 entfernt:",
      rows_before_step2 - rows_after_step2)

print("Schritt 3 - Werte 101 bis 999 entfernt:",
      rows_before_step3 - rows_after_step3)

print("Schritt 4 - Geburtsjahre umgerechnet: 0 Zeilen gelöscht")

print("\nInsgesamt gelöscht:",
      rows_before - len(df))

print("Verbleibende Zeilen:",
      len(df))
```

# **Kurze Zusammenfassung:**

Die Spalten age, hand, gender und target werden von der Prüfung ausgeschlossen. Danach werden alle Zeilen gesucht, in denen alle übrigen Spalten den Wert 0 enthalten. Diese Zeilen werden angezeigt und anschließend vollständig aus dem Datensatz entfernt.

```python
# Spalten ausschließen, die bei der Prüfung nicht berücksichtigt werden sollen
zero_columns = df.drop(columns=["age", "hand", "gender", "target"])

# Zeilen finden, in denen alle übrigen Spalten = 0 sind
zero_mask = (zero_columns == 0).all(axis=1)

# Anzahl der gefundenen Zeilen anzeigen
print("Anzahl kompletter 0-Zeilen:", zero_mask.sum())

# Gefundene Zeilen anzeigen
display(df.loc[zero_mask])

# Gefundene Zeilen entfernen
df = df.loc[~zero_mask].copy()

# Kontrolle

print("Verbleibende Zeilen im Datensatz:", len(df))

zero_mask = (zero_columns == 0).all(axis=1)
```

# 0 Prüfung
Es wird geprüft, ob in jeder Zeile mindestens ein Wert 0 vorkommt. Dabei ist egal, ob die 0 einmal oder mehrfach in derselben Zeile erscheint. Wenn dies der Fall ist, sollen auch diese Zeilen gelöscht werden, da die Daten nicht vollständig wären.

```python
# Prüfen, ob in einer Zeile mindestens eine 0 vorkommt
zero_in_row = (df == 0).any(axis=1)

print("Zeilen mit mindestens einer 0:", zero_in_row.sum())
```

```python
# Ursprüngliche, noch nicht bereinigte CSV laden
df_summary = pd.read_csv("data/data.csv")
original_rows = len(df_summary)
steps = []

# 1. Ungültige Alterswerte entfernen
before = len(df_summary)
df_summary = df_summary[df_summary["age"] <= 2002].copy()
df_summary = df_summary[~df_summary["age"].between(101, 999)].copy()
steps.append(("Ungültige age-Werte entfernt", before - len(df_summary), len(df_summary)))

# Geburtsjahre umrechnen – dabei werden keine Zeilen entfernt
birth_years = df_summary["age"].between(1900, 2002)
df_summary.loc[birth_years, "age"] = 2017 - df_summary.loc[birth_years, "age"]

# 2. Fehlende gender-Werte und 0 entfernen
before = len(df_summary)
df_summary = df_summary[
    df_summary["gender"].notna() & (df_summary["gender"] != 0)
].copy()
steps.append(("gender fehlt oder ist 0", before - len(df_summary), len(df_summary)))

# 3. Fehlende hand-Werte und 0 entfernen
before = len(df_summary)

df_summary = df_summary[
    df_summary["hand"].notna() &
    (df_summary["hand"] != 0)
].copy()

steps.append(
    (
        "hand fehlt oder ist 0",
        before - len(df_summary),
        len(df_summary)
    )
)


# 4. Zeilen entfernen, in denen alle Frage-Items 0 sind
before = len(df_summary)
zero_columns = df_summary.drop(columns=["age", "hand", "gender", "target"])
zero_mask = (zero_columns == 0).all(axis=1)
df_summary = df_summary.loc[~zero_mask].copy()
steps.append(("Alle Frage-Items sind 0", before - len(df_summary), len(df_summary)))

# Kompakte Übersicht
summary = pd.DataFrame(
    steps,
    columns=["Bereinigungsschritt", "Entfernte Zeilen", "Verbleibende Zeilen"]
)
display(summary)

print(f"Zeilen ursprünglich: {original_rows:,}")
print(f"Zeilen nach Bereinigung: {len(df_summary):,}")
```

## Eine kurze Betrachtung der Spalten und deren Werte.


### target Spalte - Verteilung der Personality Types

```python
plt.figure(figsize=(9, 5))

ax = sns.countplot(
    data=df_clean,
    x="target",
    order=df_clean["target"].value_counts().index,
    color="steelblue"
)

# Anzahl über jedem Balken anzeigen
for container in ax.containers:
    ax.bar_label(container)

plt.title("Verteilung der Persönlichkeitstypen")
plt.xlabel("Persönlichkeitstyp")
plt.ylabel("Anzahl der Personen")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

Moderate ist mit Abstand die häufigste Persönlichkeitklasse, gefolgt von Resilient. Overcontroller und Undercontroller kommen deutlich seltener vor. Die Klassen sind damit ungleich verteilt; Reserved ist in dieser Grafik nicht zu sehen.

```python
df.groupby("target").mean(numeric_only=True).round(2)

target_means = df.groupby("target").mean(numeric_only=True).round(2)
target_means = target_means.drop(columns=["age", "gender", "hand"], errors="ignore")

plt.figure(figsize=(14, 5))

sns.heatmap(
    target_means,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Durchschnitt je Persönlichkeitstyp")
plt.xlabel("Frage")
plt.ylabel("Persönlichkeitstyp")
plt.tight_layout()
plt.show()
```

Overcontroller: Hohe Zustimmung zu Stress, Sorgen und Stimmungsschwankungen; geringe Zustimmung zu geselligem Verhalten. Das spricht für hohen Neurotizismus und geringe Extraversion.
Resilient: Niedrige Werte bei Stress und Stimmungsschwankungen, höhere Werte bei Entspannung und sozialem Wohlbefinden.
Undercontroller: Höhere Zustimmung zu C4 („I make a mess of things“) und die niedrigste Zustimmung zu A4 („I sympathize with others' feelings“).
Moderate: Liegt bei vielen Fragen zwischen den anderen Gruppen.

### age, gender und target - Verteilung des Alters und Geschlechts zum Persönlichkeitstyp

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 5))

# Links: Alter nach Geschlecht
sns.boxenplot(
    data=df_clean,
    x="gender",
    y="age",
    order=["Male", "Female", "Other"],
    color="steelblue",
    ax=axes[0]
)
axes[0].set_title("Alter nach Geschlecht")
axes[0].set_xlabel("Geschlecht")
axes[0].set_ylabel("Alter in Jahren")

# Rechts: Alter nach Persönlichkeitstyp
sns.boxenplot(
    data=df_clean,
    x="target",
    y="age",
    order=["Moderate", "Resilient", "Overcontroller",
           "Undercontroller", "Reserved"],
    color="steelblue",
    ax=axes[1]
)
axes[1].set_title("Alter nach Persönlichkeitstyp")
axes[1].set_xlabel("Persönlichkeitstyp")
axes[1].set_ylabel("Alter in Jahren")
axes[1].tick_params(axis="x", rotation=25)

plt.tight_layout()
plt.show()
```

Nach Geschlecht sind die Altersverteilungen von Male und Female sehr ähnlich. Bei Other liegt der Median etwas niedriger. Nach Persönlichkeitstyp hat Resilient den höchsten Median und die größte Streuung; Overcontroller und Undercontroller wirken im Vergleich etwas jünger. In allen Gruppen gibt es einzelne deutlich ältere Personen. Für Reserved ist keine Box sichtbar, weil dieser Typ im dargestellten Datensatz nicht vorkommt.

### Antwortverteilung nach Persönlichkeitsfragen

```python
# Frage-Spalten auswählen
item_columns = df_clean.drop(
    columns=["age", "gender", "hand", "target"]
).columns

# Antworten 1 bis 5 je Frage zählen
answer_counts = pd.DataFrame({
    question: df_clean[question].value_counts().reindex(
        [1, 2, 3, 4, 5], fill_value=0
    )
    for question in item_columns
}).T

# Jede Frage einzeln auf 100 % umrechnen
answer_percent = answer_counts.div(
    answer_counts.sum(axis=1), axis=0
) * 100

plt.figure(figsize=(16, 6))

sns.heatmap(
    answer_percent.T,
    annot=True,
    fmt=".1f",
    cmap="coolwarm",
    cbar_kws={"label": "Anteil der Antworten in %"}
)

plt.title("Antwortverteilung je Frage")
plt.xlabel("Frage")
plt.ylabel("Antwort: 1 = Ablehnung bis 5 = Zustimmung")
plt.tight_layout()
plt.show()
```

Die Antworten unterscheiden sich je nach Frage deutlich. Bei N3 („I worry about things“) und A4 („I sympathize with others' feelings“) überwiegen die Zustimmungen 4 und 5. Bei C4 („I make a mess of things“) und E1 („I am the life of the party“) werden niedrige Werte häufiger gewählt. N7 und N8 haben trotz ähnlicher Formulierung keine identische Verteilung: Bei N7 wird häufiger 4 oder 5 gewählt als bei N8.

```python
data.isna().any(axis=1).sum()
df = data[-data.isna().any(axis=1)]

df.isna().sum().sum()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# df = dein DataFrame

# 1) Nur numerische Spalten
corr = df.select_dtypes(include="number").corr()

# 2) Heatmap (obere Dreiecksmatrix ausblenden)
mask = np.triu(np.ones_like(corr, dtype=bool))

plt.figure(figsize=(14, 11))
sns.heatmap(
    corr,
    mask=mask,
    annot=True,
    fmt=".2f",
    annot_kws={"size": 7},
    cmap="coolwarm",
    vmin=-1, vmax=1, center=0,
    square=True,
    linewidths=0.5,
    cbar_kws={"shrink": 0.8},
)
plt.title("Korrelationsmatrix", fontsize=14)
plt.tight_layout()
plt.show()
```
